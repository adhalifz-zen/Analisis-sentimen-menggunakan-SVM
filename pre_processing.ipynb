pre processing

```python
# prompt: cleansing membersihkan data dan kata yang tidak diperlukan seperti tanda baca, simbol, username, emoticon, mention, hashtag, dan alamat web

import re

def cleansing(sent):
  # Remove URL
  sent = re.sub(r"http\S+", " ", sent)
  # Remove mentions
  sent = re.sub(r"@\S+", " ", sent)
  # Remove hashtags
  sent = re.sub(r"#\S+", " ", sent)
  # Remove newlines
  sent = re.sub(r"\n", " ", sent)
  # Remove special characters and numbers
  sent = re.sub(r"[^a-zA-Z\s]", " ", sent)
  # Remove multiple spaces
  sent = re.sub(r"\s+", " ", sent)
  # Remove leading and trailing spaces
  sent = sent.strip()
  return sent

df = df.apply(cleansing)
print(df)
```

```python
# prompt: case folding proses mengubah seluruh tekt pada data menjadi huruf kecil (lowercase) dan menghilangkan karakter selain

def case_folding(text):
  text = text.lower()
  text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
  return text

df = df.apply(case_folding)
print(df)
```

```python
# prompt: normalization proses mengubah data kata menjadi kata yang baku

import nltk
nltk.download('punkt')
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemmatizer = WordNetLemmatizer()

def normalization(text):
  tokens = word_tokenize(text)
  normalized_tokens = [lemmatizer.lemmatize(token) for token in tokens]
  normalized_text = ' '.join(normalized_tokens)
  return normalized_text

df = df.apply(normalization)
print(df)
```

```python
# prompt: stopword proses menghilangkan kata kata yang tidak efektif atau tidak bermakna pada sebuah kalimat

nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('indonesian'))

def remove_stopwords(text):
  tokens = word_tokenize(text)
  filtered_tokens = [token for token in tokens if token not in stop_words]
  filtered_text = ' '.join(filtered_tokens)
  return filtered_text

df = df.apply(remove_stopwords)
print(df)
```

```python
# prompt: menghapus kalimat yang sama atau duplikat pada data set

df = df.drop_duplicates()
print(df)
```

```python
# prompt: menghapus data duplikat

df = df.drop_duplicates()
print(df)

num_tweets = len(df)
print(f"Jumlah tweet dalam dataframe setelah menghapus duplikat adalah {num_tweets}.")
```
